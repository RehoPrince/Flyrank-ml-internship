## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule would be too brittle here. The capstone data shows that each content item has different metadata in `dim_content`, and each item-date pair has different observed outcomes in `fact_content_daily_performance_sample`, such as impressions, clicks, pageviews, and engagement. The same item can perform differently depending on keyword profile, traffic mix, and date.

A simple rule like “prioritize high-impression content” would miss cases where traffic is high but engagement is weak, or where lower traffic still produces strong quality signals. The ranking signal is shaped by several interacting factors, not one clean threshold. ML is better suited because it can learn those combined relationships and adapt to the data rather than forcing a single rule across all content.
