## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

DF_PATH = "../../data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(DF_PATH)

df.columns
```

```python
down_trend_count = (df["trend_direction"] == "down").sum()

print("Rows: ", df.shape[0])
print("Clients: ", df["client_id"].nunique())
print("Recent-impression down items: ", down_trend_count)
print("Recent-impression down share: ", round(down_trend_count / len(df), 3))
```

Across 30,000 content items from 32 pseudonymized clients, 16,262 items (54.2%) show a down trend.

This means their impressions were lower in the most recent 30-day window than in the previous 30-day window.
