## 3. Quick look at the data (2-3 real numbers)

```python
import pandas as pd

DF_PATH = "../../data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(DF_PATH)

df.columns
```

```python
down_trend_count = (df["trend_direction"] == "down").sum()

print("Rows: ", df.shape[0])
print("Clients: ", df["client_id"].nunique())
print("Recent-impression down items: ", down_trend_count)
print("Recent-impression down share: ", round(down_trend_count / len(df), 3))
```

## 4. First group comparisons

*Compare declining items with items that are not declining. Use medians because traffic measures are heavily skewed, and show the group sizes alongside the results.*

```python
df["declining"] = df["trend_direction"] == "down"

signals = [
    "impressions_90d",
    "clicks_90d",
    "sessions_90d",
    "content_age_days",
    "days_since_last_update",
    "word_count",
]

comparison = df.groupby("declining")[signals].median().T
comparison.columns = ["not_declining_median", "declining_median"]
comparison["difference"] = (
    comparison["declining_median"] - comparison["not_declining_median"]
)

print("Group sizes:")
print(df["declining"].value_counts().rename({False: "not_declining", True: "declining"}))
print("\nMedian comparison:")
print(comparison.round(1))
```

The declining group has a higher median `impressions_90d` (961 versus 472) and is younger by median `content_age_days` (216 versus 287 days). Median clicks, sessions, update recency, and word count are very similar between groups.

These are descriptive differences, not causal evidence. They also do not establish that a signal is reliable for prioritization: the comparison uses no minimum impression floor, and missing values are omitted from each median. The next step would be to test whether these patterns hold after controlling for client and content type.
